# PICO questionnaire: auditable aggregate analysis

15 researcher-transcribed paper responses; PICO, new blue-suit/Nailong avatars, fixed order N → E → A → S. Q29 contains researcher summaries, not verbatim quotes. See [the complete Chinese report](问卷结果与完整分析.md). This notebook defaults to reading committed aggregates and never displays individual records.


```python
from pathlib import Path
import json
import sys
import subprocess

ROOT = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "Tools/AppearanceExperiment/analyze_wjx_results.py").is_file())
RESULT = ROOT / "Docs/AppearanceExperiment/Results/2026-10-21"
r = json.loads((RESULT / "analysis.json").read_text())
print("Participants:", r["included_n"], "| Valid ratings:", r["quality"]["valid_rating_cells"])
print("Order:", " → ".join(r["order"]))
print("Source SHA-256:", r["source"]["sha256"])
```

## Coverage and limits

Questionnaire formal/consent labels determine inclusion. Runtime phase/profile, APK fingerprint, trial logs and original paper sheets are not linked. Fixed order confounds condition with period; subjective ratings cannot establish causal strategy effects, task accuracy or completion time. The seven empty open questions do not mean “no problems.”


```python
for key in ("total_records", "included_n", "exclusions", "categorical_counts", "quality"):
    print(key, json.dumps(r[key], ensure_ascii=False, indent=2))
```

## Ratings

Means are descriptive; retain the ordinal frequency distributions. Interruption and effort are better when lower; control and satisfaction are better when higher. These are four separate single items, not a validated combined scale.


```python
print("Condition | Item | Question | Mean | SD | Counts 1..7")
for condition in r["order"]:
    for metric, item in r["descriptive"][condition].items():
        print(f'{condition} | {metric} | Q{item["question"]} | {item["mean"]:.3f} | {item["sd"]:.3f} | {item["counts"]}')
```

## Paired contrasts

Two-sided exact sign tests exclude ties from the binomial denominator. Holm correction covers all 12 contrasts. Mean-difference percentile intervals use 20,000 participant resamples and are unadjusted. Omnibus permutation tests are a separate four-test exploratory family; they do not remove order confounding.


```python
for item in r["pairwise"]:
    print(item["metric"], item["comparison"], "better/tied/worse:", (item["better"], item["ties"], item["worse"]), "mean difference:", round(item["mean_difference"], 4), "95% CI:", item["mean_difference_ci95"], "Holm p:", item["p_holm"])
print("Exploratory omnibus results:", json.dumps(r["friedman"], indent=2))
```

## Optional local reproduction

Requires the authorized private source workbook and the pinned analysis environment from `Tools/AppearanceExperiment/requirements-survey-analysis.txt`. Set the flag below to True only when those inputs are available. Recomputed aggregates and row audit remain in ignored StudyData; the committed results are not overwritten. The default run above reviews saved aggregates and is not a raw-data recomputation.


```python
RECOMPUTE_FROM_PRIVATE_SOURCE = False
if RECOMPUTE_FROM_PRIVATE_SOURCE:
    source = ROOT / "Docs/AppearanceExperiment/Midterm/389439995_按序号_虚拟角色外形修改研究（全流程）_15_15.xlsx"
    target = ROOT / "StudyData/survey-20261021/notebook-reproduction"
    assert source.is_file(), "Authorized source workbook is required"
    subprocess.run([sys.executable, str(ROOT / "Tools/AppearanceExperiment/analyze_wjx_results.py"), "--input", str(source), "--out", str(target), "--private-out", str(target / "private"), "--order", "N", "E", "A", "S", "--order-source", r["order_source"], "--qualitative-source", "researcher-summary", "--context-json", str(RESULT / "source-context.json"), "--no-plots"], check=True)
    reproduced = json.loads((target / "analysis.json").read_text())
    for key in ("descriptive", "pairwise", "friedman", "preferences", "qualitative", "quality"):
        assert reproduced[key] == r[key], key
    subprocess.run([sys.executable, str(ROOT / "Tools/AppearanceExperiment/validate_wjx_results.py"), "--input", str(source), "--analysis", str(target / "analysis.json"), "--out", str(target / "validation.json")], check=True)
    print("Recomputed numeric results match; independent audit passed.")
else:
    print("Recomputation disabled; displayed results are the committed aggregate snapshot.")
```
